```python
########################################
# Apprentissage non supervisé
########################################

# PCA algorithm implementation(Principal Component Analysis) in Python

# K-means clusturing  algorithm implementation in Python : 
# minimize the sum of squared distances between data points and their corresponding cluster centroids.

from sklearn.datasets import make_blobs
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from matplotlib import pyplot as plt

# Generate synthetic data
X,y= make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)
# make_blobs : generate isotropic Gaussian blobs for clustering
# centers=4 : number of centers to generate
# cluster_std=0.60 : standard deviation of the clusters
plt.scatter(X[:,0],X[:,1],s=30, cmap='viridis')

```

```python
model=KMeans(n_clusters=3, max_iter=100, n_init=10, random_state=0)
# n_clusters=3 : number of clusters to form
# max_iter=100 : maximum number of iterations for a single run
#  n_init=10 : number of time the k-means algorithm will be run with different centroid seeds

model.fit(X) # fit the model to the data
labels=model.predict(X) # predict the cluster labels for the data points
plt.scatter(X[:,0],X[:,1],c=labels, s=30, cmap='viridis')
 # plot the data points with their cluster labels
 # c=labels : color the points according to their cluster labels
plt.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1], c='red', s=35, alpha=0.5)
 # plot the cluster centers
 # s=200 : size of the points
 # alpha=0.5 : transparency of the points

print(f"Cluster centers:\n{model.cluster_centers_}") # get the cluster centers
print(f"Cluster labels:\n{model.labels_}") # get the cluster labels
```

```python
model.inertia_ # get the sum of squared distances of samples to their closest cluster center
```

```python
model.score(X) # get the opposite of the value of the inertia criterion for the model
```

```python
inertia =[]
K=range(1,20)
for k in K:
    model=KMeans(n_clusters=k, max_iter=100, n_init=10, random_state=0)
    model.fit(X)
    inertia.append(model.inertia_)

plt.plot(K,inertia,'bx-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.show()
```

```python
###########################################################################
#Isolation Forest algorithm implementation in Python:
#   Anomaly detection algorithm that identifies outliers in a dataset
#   by isolating observations through random partitioning.
###########################################################################

import numpy as np


X,y = make_blobs(n_samples=50, centers=1, cluster_std=0.50, random_state=0)

X[-1,: ]= np.array([2.25, 10])
 # add an outlier to the dataset with coordinates (2.25,3) 
 # x[-1,:] : select the last row of the array X and all columns
plt.scatter(X[:,0],X[:,1])
```

```python
from sklearn.ensemble import IsolationForest

model=IsolationForest(n_estimators=100, contamination=0.1
                      , random_state=0)
model.fit(X)
```

```python
plt.scatter(X[:,0],X[:,1],c=model.predict(X), cmap='coolwarm')
```

```python

# Importer le dataset Digits de Scikit-Learn
from sklearn.datasets import load_digits

# Charger le jeu de données des chiffres manuscrits
digits = load_digits()

# Récupérer les images des chiffres
images = digits.images

# Récupérer les données sous forme de tableau
X = digits.data

# Récupérer les classes correspondant aux chiffres
y = digits.target

# Afficher la dimension des données
print(X.shape)
```

```python
model = IsolationForest(n_estimators=100, contamination=0.1, random_state=0)
model.fit(X)
outliers = model.predict(X) == -1  # Les points considérés comme des anomalies sont marqués par -1  
print(outliers)
print(images[outliers])
```

```python
print(images[outliers])
plt.imshow(images[outliers][0])
plt.title(f"Anomalous Digit: {y[outliers][0]}")
```

```python
# PCA (Principal Component Analysis) algorithm implementation in Python:
# reduction of the dimensionality of the data while preserving as much variance as possible.
# Analyze the digits dataset using PCA to reduce its dimensionality and visualize the results.
# covariance matrix : a square matrix that contains the covariances between pairs of variables 
# in a dataset.
#Proper eigenvector



from sklearn.decomposition import PCA


model = PCA(n_components=2)  # Reduce the data to 2 dimensions
model.fit_transform(X)  # Fit the PCA model and transform the data
X.shape
```

```python
model_pca = PCA(n_components=2) # create a PCA model with 2 components
X_pca = model_pca.fit_transform(X) # fit the PCA model to the data
plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap='viridis') # plot the PCA-transformed data
plt.colorbar()
```

```python
print(f'Components: {model.components_}')
print(f'dim: {model.components_.shape}')
```

```python
#Compression of the data using PCA to reduce its dimensionality while preserving as 
# much variance as possible.


model_pca = PCA(n_components=64) # create a PCA model with 2 components
X_pca = model_pca.fit_transform(X) # fit the PCA model to the data
print(f'Explained Variance Ratio: {model_pca.explained_variance_ratio_}')
# get the explained variance ratio for each principal component

print(f'Summed Explained Variance Ratio: {np.sum(model_pca.explained_variance_ratio_)}')
#get the sum of the explained variance ratio for all principal components

plt.plot(np.cumsum(model_pca.explained_variance_ratio_))
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA Explained Variance')
plt.show()

plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap='viridis') # plot the PCA-transformed data
plt.colorbar()
X_reduced = model.fit_transform (X)
```

```python
np.argmax(np.cumsum(model_pca.explained_variance_ratio_) >= 0.99)
```

```python

# reduce the dimensionality of the data to 40 components while preserving 
# as much variance as possible.

model= PCA(n_components=40) # create a PCA model with 2 components
X_pca = model.fit_transform(X) # fit the PCA model to the data
print(f'Explained Variance Ratio: {model.explained_variance_ratio_}')
```

```python
X_reconstructed = model.inverse_transform(X_pca)
plt.imshow(X_reconstructed[0].reshape(8, 8))
```
