```python
# Charger le jeu de données Titanic avec Seaborn
titanic = sns.load_dataset('titanic')

# Afficher les premières lignes du DataFrame
titanic.head()
```

```python
# Supprimer les colonnes inutiles du DataFrame Titanic
# axis=1 indique qu'on supprime des colonnes
# inplace=True applique directement la modification à titanic

titanic.drop(
    ['alone', 'alive', 'adult_male', 'embark_town', 'class'],
    axis=1,
    inplace=True
)
# Supprimer les lignes contenant des valeurs manquantes
titanic.dropna(inplace=True)

# Afficher les premières lignes du DataFrame
titanic.head()
```

```python
sns.pairplot(titanic)
```

```python
# Créer un graphique catégoriel montrant l'âge selon la classe des passagers
sns.catplot(
    x='pclass',
    y='age',
    data=titanic,
    hue = 'sex'
)
```

```python
# Créer un graphique catégoriel montrant l'âge selon la classe des passagers
sns.boxenplot(
    x='pclass',
    y='age',
    data=titanic,
    hue = 'sex'
)
```

```python
# Afficher la distribution des tarifs payés par les passagers
sns.distplot(titanic['fare'])
```

```python
# Afficher la relation entre l'âge et le prix du billet
sns.jointplot(x='age', y='fare', data=titanic)
```

```python
# Afficher la relation entre l'âge et le prix du billet with karnel mdensity estimation
sns.jointplot(x='age', y='fare', data=titanic, kind='kde')
```

```python
# heat map avec la matrice de correlation


# Calculer la matrice de corrélation uniquement avec les colonnes numériques
correlation = titanic.corr(numeric_only=True)

# Afficher la matrice sous forme de heatmap
sns.heatmap(correlation)

plt.show()
```

```python
# Calculer la matrice de corrélation
correlation = titanic.corr(numeric_only=True)

# Créer la heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Matrice de corrélation du Titanic')
plt.show()# Calculer la matrice de corrélation
correlation = titanic.corr(numeric_only=True)

# Créer la heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Matrice de corrélation du Titanic')
plt.show()
```
