```python
# ============================================================
# Arithmetic Operations
# ============================================================

import numpy as np
np.random.seed(0)
A = np.random.randint(1, 10, (5, 5))
B = np.random.randint(1,10, (5,5))
print(A)
print(B)
print(f'Sum of A: {A.sum()}')
print(f'Sum of A along axis 0: {A.sum(axis=0)}')
print(f'Sum of A along axis 1: {A.sum(axis=1)}')
print(f'div int {A//2}')
print(f'A+B={A+B}')
print(f'A-B={A-B}')
print(f'Minimum of A: {A.min()}')
print(f'Maximum of A: {A.max()}')
print(f'Minimum of A along axis 0: {A.min(axis=0)}')
print(f'Maximum of A along axis 1: {A.max(axis=1)}')
print(f'Exponential of A: {np.exp(A)}')
print(f'Square root of A: {np.sqrt(A)}')
print(f'Square of A: {np.square(A)}')
```

```python
# ============================================================
# Statistical Operations
# ============================================================
print(f'Mean of A: {A.mean()}')
print(f'Standard deviation of A: {A.std()}') # standard deviation
print(f'Variance of A: {A.var()}') # variance
print(f'Correlation of A: {np.corrcoef(A)}')
values, count=np.unique(A, return_counts=True) # unique values with counts
print(f'valeurs uniques de A: {values}')
print(f'fréquences de A: {count}')
values[count.argsort()] # most frequent value
print(f'valeur la plus fréquente de A: {values}')
```

```python
# ============================================================
# Algebre lineaire
# ============================================================

A=np.ones((2, 3))
B=np.ones((3, 2))
print(A)
print(B)
print(f'Transpose of A: {A.T}')
print(f'Product of A and B: {A.dot(B)}') # produit matriciel




```

```python
A=np.random.randint(1, 10, (3, 3))
print(A)
print(f'determinant of an array: {np.linalg.det(A)}') # determinant

print(f'Inverse of a matrix: {np.linalg.inv(A)}') # inverse
print(f'pseudo-inverse of a matrix: {np.linalg.pinv(A)}')
print(f'Compute the eigenvalues: {np.linalg.eig(A)}') # valeurs et vecteurs propres
```

```python

C=np.random.randn(5,5)
C[0,2]=np.nan
C[4,3]=np.nan
print(C)
print(f'mean with nan :{np.nanmean(C)}')
print(f'mask of nan in the matrix:{np.isnan(C)}')
print (f'precentage of nan sum in the dataset:{np.isnan(C).sum()/C.size}')
print(C)
C[np.isnan(C)]=0
print(f'Remplace the nan per 0 :{C}')
```
