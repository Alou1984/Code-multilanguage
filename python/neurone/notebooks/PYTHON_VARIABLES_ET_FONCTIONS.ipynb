```python
#Les variables et operations de base en python

x = 1
y = 2.5

# Arithmétique
print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x ** y)

# Comparaison
print(x <= y)
print(x >= y)
print(x == y)
print(x != y)

# Logique
print(False & True)  # AND
print(False | True)  # OR
print(False ^ True)  # XOR
```
