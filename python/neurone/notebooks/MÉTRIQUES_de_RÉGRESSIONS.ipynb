```python
############################################
#  les Metriques MeanAE, MSE,MRSE, Median AR
############################################

# Importer NumPy
import numpy as np

# Importer Matplotlib
import matplotlib.pyplot as plt

# Importer les métriques d'évaluation
from sklearn.metrics import *

# Valeurs réelles
y = np.array([1, 2])

# Valeurs prédites par le modèle
y_pred = np.array([5, 2])

# Calculer et afficher la MAE (Mean Absolute Error)
print('Meam AE :', mean_absolute_error(y, y_pred))

# Calculer et afficher la MSE (Mean Squared Error)
print('MSE :', mean_squared_error(y, y_pred))

# Calculer et afficher la RMSE (Root Mean Squared Error)
print('RMSE :', np.sqrt(mean_squared_error(y, y_pred)))

# Calculer et afficher la RMSE (Median Absolute Error)
print('Median AE :', np.sqrt(median_absolute_error(y, y_pred)))
```

```python
# Importer le jeu de données Boston
import pandas as pd

# Importer le modèle de régression linéaire
from sklearn.linear_model import LinearRegression

# Charger les données Boston
boston = pd.read_csv('boston.csv')
#boston = load_boston()

# Variables explicatives
# On supprime la colonne medv car c'est la variable cible
X = boston.drop('medv', axis=1)

# Variable cible : prix des maisons
y = boston['medv']

# Créer le modèle de régression linéaire
model = LinearRegression()

# Entraîner le modèle avec X et y
model.fit(X, y)

# Faire les prédictions
y_pred = model.predict(X)

# Afficher les valeurs réelles en fonction de la 6e variable
# X.iloc[:, 5] correspond à la 6e colonne : RM
plt.scatter(X.iloc[:, 5], y, label='y')

# Afficher les valeurs prédites
plt.scatter(X.iloc[:, 5], y_pred, alpha=0.8, label='y_pred')

# Afficher la légende
plt.legend()

# Afficher le graphique
plt.show()
```

```python
# Calculer l'erreur entre les valeurs réelles et les valeurs prédites
err_hist = np.abs(y - y_pred)

# Afficher l'histogramme des erreurs
plt.hist(err_hist, bins=50)

# Afficher le graphique
plt.show()
```

```python
# Calculer et afficher la MAE (Mean Absolute Error)
print('MAE :', mean_absolute_error(y, y_pred))

# Calculer et afficher la MSE (Mean Squared Error)
print('MSE :', mean_squared_error(y, y_pred))

# Calculer et afficher la RMSE (Root Mean Squared Error)
print('RMSE :', np.sqrt(mean_squared_error(y, y_pred)))

# Calculer et afficher la RMSE (Median Absolute Error)
print('MAE :', np.sqrt(median_absolute_error(y, y_pred)))
```

```python
############################################
# Le Coefficient de determination R carre -> Methode Score
############################################


print(f'Score = {model.score(X,y)}')



```
