```python

# ============================================================
# Les graphes les plus importants
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

x = iris.data
y = iris.target
names = list(iris.target_names)

print(f'x contient {x.shape[0]} exemples et {x.shape[1]} variables')
print(f'il y a {np.unique(y).size} classes')
print(f'les entrees {x}')
print(f'les sorties {y}')
```

```python
# ============================================================
# Exercice
# ============================================================


dataset = {"experience {}".format(i): np.random.randn(100) for i in range(7)}

# dataset

def graphique(data):
    n = len(data)
    plt.figure(figsize=(12, 8))

    for i, k in enumerate(data.keys(), start=1):
        plt.subplot(n, 1, i)
        plt.plot(data[k])
        plt.title(k)

    plt.show()

```

```python
graphique(dataset)
```

```python
f = lambda x, y: np.sin(x) + np.cos(x + y)

X = np.linspace(0, 5, 100)
Y = np.linspace(0, 5, 100)
X, Y = np.meshgrid(X, Y)
Z = f(X, Y)
print(Z.shape)
ax = plt.axes(projection='3d')
ax.plot_surface(X,Y,Z,cmap='plasma')
```

```python
# ============================================================
# Analyse d'image avec un Histogramme
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

image = plt.imread("image.png")

# Si l'image est en couleur, la convertir en niveaux de gris
if image.ndim == 3:
    image = np.mean(image[:, :, :3], axis=2)

# Redimensionnement par sélection de pixels
image_255 = image[
    np.linspace(0, image.shape[0] - 1, 255).astype(int),
    :
]

image_255 = image_255[
    :,
    np.linspace(0, image.shape[1] - 1, 255).astype(int)
]

print("Image originale :", image.shape)
print("Image réduite :", image_255.shape)

#plt.imshow(image_255, cmap='gray')
plt.hist(image_255.ravel(),bins=255)
plt.axis('off')
plt.show()



```

```python
# Récupérer les différentes classes présentes dans y
classes = list(np.unique(y))

# Afficher les classes
print(classes)

# Nombre de variables/features dans X
n = x.shape[1]

# Créer une figure
plt.figure(figsize=(12, 8))

# Parcourir toutes les variables
for i in range(n):
    # Créer un sous-graphe
    plt.subplot(n//2, n//2, i+1)

    # Afficher les données avec les classes représentées par les couleurs
    plt.scatter(x[:, 0], x[:, i], c=y)

    # Nom de l'axe horizontal
    plt.xlabel('0')

    # Nom de l'axe vertical
    plt.ylabel(i)

    # Ajouter une barre de couleurs avec les classes
    plt.colorbar(ticks=list(np.unique(y)))

# Afficher la figure
plt.show()
```
