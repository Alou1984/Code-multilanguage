```python
# ============================================================
# Standardiser la matrice  A sur chaque colonne 
# Chaque colonne doit avoir :
# - Une moyenne 0
# - un ecart type de 1
# ========================================================
import random
import numpy as np


np.random.seed(0)
A = np.random.randint(0, 100, [10, 5])
print(A)
D = (A- A.mean(axis=0))/A.std(axis=0)
print(f'Standardisation de A par colonne: {D}')
print(f'La moyenne est null ou proche de zero:{D.mean(axis=0)}')
print(f'L ecart typede chaque colonne est de 1{D.std(axis=0)}')
```
