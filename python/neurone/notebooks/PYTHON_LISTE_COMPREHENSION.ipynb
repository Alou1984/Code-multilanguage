```python
# tous les carrés de 0 à 9
liste_1 = []

for i in range(10):
    liste_1.append(i**2)

print(liste_1)
```

```python
# ============================================================
# Les compréhensions de liste sont une manière plus concise de créer des listes
# ============================================================

liste_2 = [i**2 for i in range(10)]
print(liste_2)  
```

```python
import time

start = time.time()

liste_1 = []
for i in range(10000000):
    liste_1.append(i**2)

end = time.time()
print(end - start)


start = time.time()

liste_2 = [i**2 for i in range(10000000)]

end = time.time()
print(end - start)
```

```python
liste_3 = [[i+j for i in range(2)] for j in range(4)]

print(liste_3)
```

```python
prenoms = ['Pierre', 'Jean', 'Julie', 'Sophie']

dico = {k: v for k, v in enumerate(prenoms)}
print(dico)

dico.values()
print(dico.values())
```

```python
ages = [24, 62, 10, 23]

dico_2 = {prenom: age for prenom, age in zip(prenoms, ages) if age > 20 }
print(dico_2)
```
