```python
# ============================================================
# Pyplot
# ============================================================

import matplotlib.pyplot as plt
import numpy as np
plt.figure()
x=np.linspace(0,2.10)
y= x**2
print(x)
plt.plot(x, y,label='Quadratic', c='green',lw=5, ls='--')
plt.plot(x, x**3, label='Cubique')
plt.scatter(x,y)
plt.title('figure 1')
plt.xlabel('axe x')
plt.ylabel('axe y')
plt.legend()
```

```python
plt.subplot(2, 1, 1)
plt.plot(x, y, c='red')

plt.subplot(2, 1, 2)
plt.plot(x, y, c='blue')
plt.plot(x, np.sin(x), c='green')
```

```python
# ============================================================
# Pyplot with POOmmethod
# ============================================================



fig, ax = plt.subplots(2, 1, sharex=True)

ax[0].plot(x, y)
ax[1].plot(x, np.sin(x))

plt.show()
```
