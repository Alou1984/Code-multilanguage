```python
#############################################
# Cross validation avec Kfold
#############################################


# Importer NumPy
import numpy as np

# Importer Matplotlib pour les graphiques
import matplotlib.pyplot as plt

# Importer le jeu de données Iris
from sklearn.datasets import load_iris


# Charger les données Iris
iris = load_iris()

# Variables explicatives
X = iris.data

# Variable cible (classes)
y = iris.target


# Afficher la dimension de X
print(X.shape)

# Afficher un nuage de points avec les deux premières variables
# La couleur dépend de la classe y
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y,
    alpha=0.8
)

plt.show()


# Importer la validation croisée
from sklearn.model_selection import cross_val_score

# Importer KFold pour créer les différents plis
from sklearn.model_selection import KFold

# Importer le model KNN 
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Créer une validation croisée en 5 parties (K-Fold)
# random_state=0 permet de garder toujours la même séparation
cv = KFold(5,shuffle=True, random_state=0)

# Évaluer le modèle KNN avec la validation croisée
cross_val_score(
    KNeighborsClassifier(),
    X,
    y,
    cv=cv
)
```

```python
#############################################
# Cross validation avec leave one out 
#############################################

# Importer KFold pour créer les différents plis
from sklearn.model_selection import LeaveOneOut

# Créer une validation croisée en  choisisant un echamtillon comme un groupe 
# random_state=0 permet de garder toujours la même séparation
cv = LeaveOneOut()

# Évaluer le modèle KNN avec la validation croisée
cross_val_score(
    KNeighborsClassifier(),
    X,
    y,
    cv=cv
)
```

```python
#############################################
# Cross validation avec shuffle plit
#############################################

# Importer Shuffle pour créer les différents plis
from sklearn.model_selection import ShuffleSplit

# Créer une validation croisée en  choisisant un echamtillon d 'entrainement comme groupe 
# random_state=0 permet de garder toujours la même séparation
cv = ShuffleSplit(4, test_size=0.2)

# Évaluer le modèle KNN avec la validation croisée
cross_val_score(
    KNeighborsClassifier(),
    X,
    y,
    cv=cv
)
```

```python
#############################################
# Cross valid Stratified 
#############################################

from sklearn.model_selection import StratifiedKFold

# Créer une validation croisée en  choisisant un echamtillon d 'entrainement comme groupe 
# random_state=0 permet de garder toujours la même séparation
cv = StratifiedKFold(4)

# Évaluer le modèle KNN avec la validation croisée
cross_val_score(
    KNeighborsClassifier(),
    X,
    y,
    cv=cv
)
```

```python
#############################################
# Cross valid Groupfold
#############################################

from sklearn.model_selection import GroupKFold

# Créer une validation croisée en  choisisant un echamtillon d 'entrainement comme groupe 
# random_state=0 permet de garder toujours la même séparation
cv = GroupKFold(5).get_n_splits(X,y, groups=X[:,0])

# Évaluer le modèle KNN avec la validation croisée
cross_val_score(
    KNeighborsClassifier(),
    X,
    y,
    cv=cv
)
```
