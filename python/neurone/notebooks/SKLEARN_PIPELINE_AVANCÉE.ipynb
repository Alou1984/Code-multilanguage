```python
# Importer NumPy pour effectuer des calculs numériques
import numpy as np

# Importer Matplotlib pour créer des graphiques
import matplotlib.pyplot as plt

# Importer make_pipeline pour créer une pipeline
# qui enchaîne plusieurs étapes de traitement
from sklearn.pipeline import make_pipeline

# Importer StandardScaler pour standardiser les données
# (moyenne = 0 et écart-type = 1)
from sklearn.preprocessing import StandardScaler

# Importer SGDClassifier pour créer un modèle
# de classification basé sur la descente de gradient stochastique
from sklearn.linear_model import SGDClassifier

# Importer Seaborn pour la visualisation et les jeux de données
import seaborn as sns


# Charger le jeu de données Titanic fourni par Seaborn
titanic = sns.load_dataset('titanic')

# Afficher les premières lignes du jeu de données
# Par défaut, head() affiche les 5 premières lignes
titanic.head()
```

```python
#############################################################################
# make_column_transformer -> pour transformer certain colonne 
#############################################################################

# y = la variable cible (target)
# 'survived' indique si le passager a survécu :
# 0 = n'a pas survécu
# 1 = a survécu
y = titanic['survived']


# X = les variables explicatives (features)
# On retire la colonne 'survived' des données
# car c'est justement la variable que le modèle doit prédire.
#
# axis=1 signifie qu'on supprime une colonne
X = titanic.drop('survived', axis=1)



from sklearn.compose  import make_column_transformer
from sklearn.impute   import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Variables numériques
numerical_features = ['pclass', 'age', 'fare']  # autres variables masquées

# Variables catégorielles
categorical_features = ['sex', 'deck', 'alone']  # éventuelles colonnes masquées


# Pipeline pour les variables numériques
numerical_pipeline = make_pipeline(
    SimpleImputer(strategy='median'),  # Remplacer les valeurs manquantes par la médiane
    StandardScaler()                   # Standardiser les variables
)


# Pipeline pour les variables catégorielles
categorical_pipeline = make_pipeline(
    SimpleImputer(strategy='most_frequent'),  # Remplacer les valeurs manquantes par la valeur la plus fréquente
    OneHotEncoder()                            # Transformer les catégories en variables numériques
)

# Appliquer chaque pipeline aux colonnes correspondantes
preprocessor  = make_column_transformer(
    
    (numerical_pipeline, numerical_features),
    (categorical_pipeline, categorical_features)
    
)

model = make_pipeline(preprocessor, SGDClassifier())
# Entraîner toute la pipeline
#
# X = variables explicatives
# y = variable cible
#
# La pipeline va automatiquement :
# 1. standardiser X avec StandardScaler
# 2. entraîner SGDClassifier sur les données standardisées

model.fit(X, y)
print("Modèle entraîné avec succès !")
```

```python
################################################################
# make_union_transformer -> pour transformer en paralelle
################################################################
from sklearn.pipeline  import make_union
from sklearn.preprocessing import Binarizer

# y = la variable cible (target)
# 'survived' indique si le passager a survécu :
# 0 = n'a pas survécu
# 1 = a survécu
y = titanic['survived']


# X = les variables explicatives (features)
# On retire la colonne 'survived' des données
# car c'est justement la variable que le modèle doit prédire.
#
# axis=1 signifie qu'on supprime une colonne
X = titanic.drop('survived', axis=1)

numerical_features = X[['age', 'fare']] # Sélectionner toutes les colonnes numériques
print("Colonnes numériques sélectionnées :", numerical_features.columns.tolist())

# Première branche :
# remplacer les valeurs manquantes par la médiane
numerical_pipeline = make_pipeline(
    SimpleImputer(strategy='median')
)

# Deuxième branche :
# remplacer les valeurs manquantes puis appliquer Binarizer
binarizer_pipeline = make_pipeline(
    SimpleImputer(strategy='median'),
    Binarizer()
)
# Appliquer les deux pipelines en parallèle
pipeline = make_union(
    numerical_pipeline,
    binarizer_pipeline
)

# Transformer les données
X_transformed = pipeline.fit_transform(numerical_features)

print(X_transformed)

X_transformed.shape
```
