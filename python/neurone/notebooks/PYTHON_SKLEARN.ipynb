```python
##########################
# LinearRegression
##########################

import numpy as np
import matplotlib.pyplot as plt

# 1) Regression avec SKLearn

np.random.seed(0)

m = 100  # création de 100 échantillons

X = np.linspace(0, 10, m).reshape(m, 1)

y = X + np.random.randn(m, 1)
plt.scatter(X,y)






```

```python
# Importer le modèle de régression linéaire de Scikit-Learn
from sklearn.linear_model import LinearRegression

# Créer le modèle
model = LinearRegression()

# Entraîner le modèle avec X et y
model.fit(X, y)

# Calculer le score R² du modèle
print(f'Score = {model.score(X, y)}')


# Faire des prédictions à partir de X
model.predict(X)
```

```python
# Créer le modèle de régression linéaire
model = LinearRegression()

# Entraîner le modèle avec X et y
model.fit(X, y)

# Calculer le score R² du modèle
model.score(X, y)

# Faire les prédictions
predictions = model.predict(X)

# Afficher les données réelles sous forme de nuage de points
plt.scatter(X, y)

# Afficher la droite de régression en rouge
plt.plot(X, predictions, c='r')
```

```python
################################################################
# KNN - K Nearest Neighbors
# Algorithme des K plus proches voisins
################################################################

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Charger les données Titanic
titanic = pd.read_excel('titanic3.xls')

# Afficher les dimensions du DataFrame
print(titanic.shape)

# Afficher les 5 premières lignes
titanic.head()

# Garder uniquement les colonnes utiles pour l'analyse
titanic = titanic[['survived', 'pclass', 'sex', 'age']].copy()

# Supprimer les lignes contenant des valeurs manquantes
titanic.dropna(axis=0, inplace=True)

# Transformer la variable 'sex' en valeurs numériques
# male   -> 0
# female -> 1
titanic['sex'] = titanic['sex'].replace(['male', 'female'], [0, 1])

# Afficher les 5 premières lignes
titanic.head()
```

```python
# Importer l'algorithme K plus proches voisins (KNN)
from sklearn.neighbors import KNeighborsClassifier

# Créer le modèle KNN
model = KNeighborsClassifier()

# Définir la variable cible : survie
y = titanic['survived']

# Créer les variables explicatives en supprimant 'survived'
X = titanic.drop('survived', axis=1)

# Afficher X
print(f'X= {X}')
# Afficher y
print(f'y= {y}')

# Entraîner le modèle avec X et y
model.fit(X, y)

# Calculer le score R² du modèle
model.score(X, y)
print(f'Score = {model.score(X, y)}')

# Faire les prédictions
#predictions = model.predict(X)

def survie(model, pclass=3, sex=0, age=26):
    # Créer les caractéristiques d'un nouveau passager
    x = np.array([pclass, sex, age]).reshape(1, 3)

    # Prédire si le passager survit ou non
    print(model.predict(x))
    print(model.predict_proba(x))


survie(model, 1, 0, 42)
```

```python
for i in range (1 , 10):
    # Créer le modèle KNN avec 5 voisins
    model = KNeighborsClassifier(n_neighbors=i)

    # Définir la variable cible : survie
    y = titanic['survived']

    # Définir les variables explicatives
    X = titanic.drop('survived', axis=1)

    # Entraîner le modèle
    model.fit(X, y)

    # Calculer le score du modèle
    print( f'score {i} ={model.score(X, y)}')
```
