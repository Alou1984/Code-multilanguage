```python
# Importer NumPy pour les calculs numériques
import numpy as np
# Importer Pandas pour manipuler les données
import pandas as pd
# Importer Matplotlib pour créer des graphiques
import matplotlib.pyplot as plt


# Charger le fichier CSV contenant les données Iris
iris = pd.read_csv('iris.csv')
#Afficher les 5 premières lignes du DataFrame
iris.head()


```

```python


# Les variables explicatives
X = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]

# Normaliser les données entre 0 et 1
X_minmax = MinMaxScaler().fit_transform(X)

# Standardiser les données entre 0 et 1
Scaler = StandardScaler().fit_transform(X)

# Afficher un nuage de points avec les colonnes 3 et 4
plt.scatter(X.iloc[:, 2], X.iloc[:, 3], label='Original')

# Afficher les colonnes 3 et 4 normalisées
plt.scatter(X_minmax[:, 2], X_minmax[:, 3], label='Normalise')

# Afficher les colonnes 3 et 4 standardise
plt.scatter(Scaler[:, 2], Scaler[:, 3], label='Standardise')

plt.legend()

# Afficher le graphique
plt.show()
```

```python
# Presence de outliers  et sensibilites

# Créer 10 valeurs aberrantes (outliers) avec 4 variables
outliers = np.full((10, 4), 100) + np.random.randn(10, 4)

# Ajouter les outliers aux données X
X = np.vstack((X, outliers))

# Normalisation Min-Max : transforme les valeurs entre 0 et 1
X_minmax = MinMaxScaler().fit_transform(X)

# Standardisation : centre les données autour de 0
# avec un écart-type égal à 1
X_stdsc = StandardScaler().fit_transform(X)

# Afficher les données originales
plt.scatter(
    X[:, 2],
    X[:, 3],
    alpha=0.5,
    label='original'
)

# Afficher les données après normalisation Min-Max
plt.scatter(
    X_minmax[:, 2],
    X_minmax[:, 3],
    alpha=0.5,
    label='min-max'
)

# Afficher les données après standardisation
plt.scatter(
    X_stdsc[:, 2],
    X_stdsc[:, 3],
    alpha=0.5,
    label='standardisation'
)

# Afficher la légende
plt.legend()

# Afficher le graphique
plt.show()
```

```python
# Robust Scaler

# Importer MinMaxScaler pour standardiser les données
from sklearn.preprocessing import RobustScaler

# Créer 10 outliers avec 4 variables
outliers = np.full((10, 4), 100) + np.random.randn(10, 4)

# Ajouter les outliers aux données X
X = np.vstack((X, outliers))

# Normalisation Min-Max
X_minmax = MinMaxScaler().fit_transform(X)

# Standardisation
X_stdsc = StandardScaler().fit_transform(X)

# Standardisation robuste aux valeurs aberrantes
X_robust = RobustScaler().fit_transform(X)

# Afficher les données originales
plt.scatter(X[:, 2], X[:, 3], alpha=0.5, label='original')

# Afficher les données après Min-Max
plt.scatter(X_minmax[:, 2], X_minmax[:, 3], alpha=0.5, label='min-max or normalisation')

# Afficher les données après StandardScaler
plt.scatter(X_stdsc[:, 2], X_stdsc[:, 3], alpha=0.5, label='standard')

# Afficher les données après RobustScaler
plt.scatter(X_robust[:, 2], X_robust[:, 3], alpha=0.5, label='robust')

# Afficher la légende
plt.legend()

# Afficher le graphique
plt.show()
```

```python

#Transformation non lineaire 

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Créer 100 valeurs entre 0 et 4
# reshape(100, 1) transforme le tableau en 100 lignes et 1 colonne
X = np.linspace(0, 4, 100).reshape((100, 1))

# Créer la variable cible y
# x² + 5*cos(x) représente une relation non linéaire
# np.random.randn() ajoute du bruit aléatoire
y = X**2 + 5*np.cos(X) + np.random.randn(100, 1)

# Afficher les données sous forme de nuage de points
plt.scatter(X, y,label='Données')

# Créer le transformateur polynomial de degré 5 et
# Transformer X en variables polynomiales
X_poly= PolynomialFeatures(5).fit_transform(X)

# Créer et entraîner un modèle de régression linéaire
model = LinearRegression().fit(X_poly, y)

# Faire les prédictions
y_pred = model.predict(X_poly)

# Afficher la droite de régression en rouge
plt.plot(X, y_pred, c='r', lw=3)


# Afficher le graphique
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PowerTransformer

# Set the random seed so that the same random numbers
# are generated every time the code is executed
np.random.seed(0)

# Generate a dataset with 100 rows and 2 columns
# np.random.rand() generates uniformly distributed values aléatoires entre 0 et 1
# np.random.randn() generates normally distributed values
X = np.random.rand(100, 2) + np.random.randn(100, 2)

# Apply PowerTransformer to transform the data
# The transformation attempts to make the data
# more normally distributed
X2 = PowerTransformer().fit_transform(X)

# Create a histogram of the first column of the original data
# bins=30 -> divide the data into 30 intervals
# alpha=0.7 -> make the histogram slightly transparent
plt.hist(X[:, 0], bins=30, alpha=0.7, label='original')

# Create a histogram of the first column after
# applying the Power Transformation
plt.hist(X2[:, 0], bins=30, alpha=0.7, label='PowerTransfo')

# Display the labels for the two histograms
plt.legend()

# Display the final plot
plt.show()
```

```python
# Pipline 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDClassifier

# Charger le fichier CSV contenant les données Iris
iris = pd.read_csv('iris.csv')
#Afficher les 5 premières lignes du DataFrame
iris.head()
 

# Les variables explicatives
X = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
#Transformer les espèces en valeurs numériques
y = iris["species"].map({
    "setosa": 0,
    "versicolor": 1,
    "virginica": 3
})

# Séparer les données en deux parties :
# X_train, y_train → données utilisées pour entraîner le modèle
# X_test, y_test   → données utilisées pour tester le modèle
X_train, X_test, y_train, y_test = train_test_split(X, y)


# ==================================================
# Transformer
# ==================================================

# Importer le StandardScaler pour standardiser les données
from sklearn.preprocessing import StandardScaler

# Créer le scaler pour standardiser les donnes 
scaler = StandardScaler()

# Calculer la moyenne et l'écart-type sur les données d'entraînement,
# puis standardiser les données d'entraînement
X_train_transformed = scaler.fit_transform(X_train)


# ==================================================
# Estimator (modèle)
# ==================================================

# Créer un modèle de classification SGD
# random_state=0 permet d'obtenir des résultats reproductibles
model = SGDClassifier(random_state=0)

# Entraîner le modèle avec les données transformées
# et les vraies classes correspondantes
model.fit(X_train_transformed, y_train)


# ==================================================
# Test
# ==================================================

# Standardiser les données de test
# IMPORTANT : on utilise uniquement le scaler déjà entraîné
# sur les données d'entraînement
X_test_transformed = scaler.transform(X_test)

# Utiliser le modèle entraîné pour prédire
# les classes des données de test
model.predict(X_test_transformed)


```

```python
# a pipline this code becomes 
#-------------------------------------------------------------------------
# Une pipeline est simple à utiliser
# Évite d’avoir des fuites de données, ou des données mal transformées
# Permet de faire des cross-validation
#--------------------------------------------------------------------------


from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDClassifier

# Créer une pipeline qui enchaîne deux étapes :
# 1. StandardScaler() → standardiser les données
# 2. SGDClassifier() → entraîner le modèle de classification
model = make_pipeline(
    StandardScaler(),
    SGDClassifier()
)

# Entraîner toute la pipeline sur les données d'entraînement
# Le StandardScaler est d'abord ajusté et transforme X_train,
# puis le SGDClassifier est entraîné sur les données transformées
model.fit(X_train, y_train)

# Faire des prédictions sur les données de test
# La pipeline applique automatiquement la même standardisation
# puis utilise le modèle pour faire les prédictions
model.predict(X_test)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import SGDClassifier

# Créer une pipeline avec 3 étapes :
model = make_pipeline(

    # 1. PolynomialFeatures :
    # Crée de nouvelles variables à partir des variables existantes
    # Exemple : x devient x², x³, et peut créer des interactions
    PolynomialFeatures(),

    # 2. StandardScaler :
    # Standardise les variables :
    # moyenne = 0 et écart-type = 1
    StandardScaler(),

    # 3. SGDClassifier :
    # Modèle de classification supervisée
    #
    # SGD = Stochastic Gradient Descent
    #       (descente de gradient stochastique)
    #
    # Il apprend à séparer les différentes classes
    # à partir de X_train et y_train.
    #
    # Par défaut :
    # loss="hinge" → modèle linéaire de type SVM
    #
    # Exemple :
    # 0 → classe "setosa"
    # 1 → classe "versicolor"
    # 3 → classe "virginica"
    SGDClassifier()
)
model
# Définir les paramètres que GridSearchCV va tester
params = {

    # Tester plusieurs degrés pour PolynomialFeatures
    # 2 → créer des termes jusqu'au degré 2
    # 3 → créer des termes jusqu'au degré 3
    # 4 → créer des termes jusqu'au degré 4
    'polynomialfeatures__degree': [2, 3, 4],

    # Tester deux types de régularisation pour SGDClassifier
    #
    # L1 → peut mettre certains coefficients exactement à 0
    #      et donc effectuer une forme de sélection de variables
    #
    # L2 → réduit généralement la taille des coefficients
    #      sans forcément les mettre à 0
    'sgdclassifier__penalty': ['l1', 'l2']
}

# Créer une recherche par grille
# GridSearchCV va tester automatiquement toutes les combinaisons
# des paramètres indiqués dans "params"
#
# Ici :
# 3 degrés × 2 pénalités = 6 combinaisons
#
# cv=5 signifie une validation croisée en 5 parties
grid = GridSearchCV(
    model,
    param_grid=params,
    cv=5
)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDClassifier


# Charger le fichier CSV contenant les données Iris
iris = pd.read_csv('iris.csv')
#Afficher les 5 premières lignes du DataFrame
iris.head()
 
# Les variables explicatives
X = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
#Transformer les espèces en valeurs numériques
y = iris["species"].map({
    "setosa": 0,
    "versicolor": 1,
    "virginica": 3
})

# Séparer les données en deux parties :
# X_train, y_train → données utilisées pour entraîner le modèle
# X_test, y_test   → données utilisées pour tester le modèle
#
# random_state=0 → permet d'obtenir toujours la même séparation
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=0
)


# Créer le modèle SGDClassifier
# random_state=0 → rend l'entraînement reproductible
model = SGDClassifier(random_state=0)

# Entraîner le modèle avec les données d'entraînement
# X_train = caractéristiques
# y_train = classes correspondantes
model.fit(X_train, y_train)

# Évaluer le modèle sur les données de test
# X_test = données que le modèle n'a pas utilisées pour apprendre
# y_test = vraies classes des données de test
#
# .score() retourne ici l'accuracy (taux de bonnes prédictions)
score = model.score(X_test, y_test)

# Afficher le score
print(score)
```
