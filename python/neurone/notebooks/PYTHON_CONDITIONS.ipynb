```python
# IF ELSE ELIF ELSE

x = 1
y = 2

if (x > 0) & (y > x):
    print(x, y, 'positif')
elif x == 0:
    print(x, 'nul')
else:
    print(x, 'negatif')
```

```python
def signe(x):
    if x > 0:
        return 'positif'
    elif x == 0:
        return 'nul'
    else:
        return 'negatif'
```

```python
a=signe(-5847)
print(a)
```

```python
#Boucle for 
for element in range(10, -10, -3):
    print(element, signe(element))
```
